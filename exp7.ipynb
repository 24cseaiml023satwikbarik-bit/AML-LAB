```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame({
    "rainfall":["high","high",'high','low','low','low'],
    "soilType":["loamy","sandy","clay","loamy","sandy","clay"],
    "yield":["good","good","good","poor","poor","good"]
})
df
```

```python
def one_r_algorithm(df, target_col):
    N = len(df)
    features = [col for col in df.columns if col != target_col]
    attribute_errors = {}
    rules = {}

    for xj in features:  # Step 1: select attribute xj
        distinct_values = df[xj].unique()  # Step 2: list distinct values
        xj_rules = {}
        incorrect_count = 0

        for v in distinct_values:
            # Step 3: count target class for each value v
            counts = df[df[xj] == v][target_col].value_counts()

            # Step 4 & 5: assign majority class (rule(v) = argmax count(v,c))
            majority_class = counts.idxmax()
            xj_rules[v] = majority_class

            # Step 6: count incorrectly classified observations
            incorrect_count += (
                df[df[xj] == v][target_col] != majority_class
            ).sum()

        # Step 6: error calculation
        error = incorrect_count / N
        attribute_errors[xj] = error
        rules[xj] = xj_rules

        print(f"Attribute: {xj}")
        print(f"  Rules: {xj_rules}")
        print(f"  Error: {error:.4f}\n")

    # Step 7: select attribute with minimum error (x* = argmin error)
    best_attribute = min(attribute_errors, key=attribute_errors.get)

    print(
        f"Selected Best Attribute (x*): {best_attribute} with Error: {attribute_errors[best_attribute]:.4f}"
    )
    print(f"Final OneR Rule Set: {rules[best_attribute]}")


one_r_algorithm(df, target_col="yield")
```
